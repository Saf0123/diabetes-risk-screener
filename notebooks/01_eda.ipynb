```python
import pandas as pd, numpy as np
print("Notebook is working")
```

```python
pd.set_option("display.max_columns", None)
df = pd.read_csv("../data/Pakistani_Diabetes_Dataset.csv")
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
print(df.dtypes)
print(df.isna().sum())
print("duplicates:", df.duplicated().sum())
print(df["Outcome"].value_counts())
df.describe(include="all").T
```

```python
print(df.groupby("Outcome")["Age"].describe())
```

```python
num = df.select_dtypes("number").columns.drop("Outcome")
print(df.groupby("Outcome")[list(num)].mean().T)
```

```python
from sklearn.metrics import roc_auc_score
d = df.dropna()
for c in num:
    try:
        print(c, round(roc_auc_score(d["Outcome"], d[c]), 3))
    except Exception as e:
        pass
```
